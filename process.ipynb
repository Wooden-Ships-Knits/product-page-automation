```python
import os
import pandas as pd
import openpyxl
path = "process/DATA_fixed.xlsx"

df = pd.read_excel(path)
```

```python
df
```

```python
path_sat = "SUMMARIZE ALL TIME.xlsx"
df_sat = pd.read_excel(path_sat, sheet_name="ALL TIME SALES")
```

```python
df_sat
```

```python
df_sat_net_sales = df_sat.groupby("Product ID")["Net sales"].sum().reset_index()
```

```python
df_sat_net_items= df_sat.groupby("Product ID")["Net items sold"].sum().reset_index()
```

```python
df_sat_check = df_sat_net_sales.merge(df_sat_net_items, on="Product ID")
```

```python
df_sat_check = df_sat_check.sort_values("Net items sold", ascending=True)
```

```python
df_sat_check
```

```python
df_p = df 
df_p = df_p.merge(df_sat_check, on="Product ID", how="left")
```

```python
df_p = df_p.dropna(subset=["Product ID"])
```

```python
# df_p = df_p.sort_values(by=['Style + Color', 'Product ID'], ascending=[True, False])
df_p
```

```python
df_p["Current Production Type"] = df_p["Style + Color"].str.strip() + "+" + df_p["FP/DC"].astype(str)
```

```python
df_p = df_p.sort_values(by=['Current Production Type','Product ID'], ascending=[True, False])
```

```python
df_p = df_p[df_p['Current Production Type'].duplicated(keep=False)]
```

```python
df_p['Net items sold'] = df_p['Net items sold'].fillna(0)
```

```python
df_p = df_p.sort_values(by=['Current Production Type','Net items sold'], ascending=[True, False])
```

```python
df_p['Net items sold'] = df_p['Net items sold'].astype(int)
```

```python
# df_final = df_p.loc[df_p['Units Sold (since 2016)'] < 6]
```

```python
df_final = df_p.sort_values(by=['Current Production Type','Net items sold'], ascending=[True, False])
```

```python
df_final
```

```python
df_final = df_final.loc[df_final['Page Status'] == "DRAFT"]
```

```python
productID_to_keep = []
style_color_to_keep = []
for i, d in df_final.iterrows():
    if d['Current Production Type'] not in style_color_to_keep:
        style_color_to_keep.append(d['Current Production Type'])
        productID_to_keep.append(d['Product ID'])
    # else:
    #     df_final.drop(i, inplace=True)
```

```python
df_final_not_keep= df_final[~df_final['Product ID'].isin(productID_to_keep)]
```

```python
df_final_not_keep = df_final_not_keep.loc[df_final_not_keep['Net items sold'] < 6]
```

```python
df_final_not_keep
```

```python
productID_to_delete = df_final_not_keep['Product ID'].tolist()
```
